```python
import random
import pandas as pd
from faker import Faker
```

## Initialize Faker

```python
fake = Faker()
```

## Define categories and sample templates

```python
CATEGORY_TEMPLATES = {
    "Stationary": [
        "Bought {item} for ${amount} at the local store",
        "Spent ${amount} on {item} and pens",
        "Purchased some {item} for office use",
        "Stationary expense: {item} costing ${amount}"
    ],
    "Electronics": [
        "Bought a new {item} for ${amount}",
        "Purchased {item} from BestBuy",
        "Electronics expense: {item} for personal use",
        "Spent ${amount} on {item} during sale"
    ],
    "Groceries": [
        "Spent ${amount} on {item} at Walmart",
        "Bought fresh {item} and vegetables",
        "Grocery shopping at {store}, ${amount}",
        "Purchased {item} and fruits for the week"
    ],
    "Travel": [
        "Booked a flight to {city} for ${amount}",
        "Spent ${amount} on train tickets to {city}",
        "Travel expense: bus tickets to {city}",
        "Trip to {city}, total cost ${amount}"
    ],
    "Food": [
        "Had lunch at {store}, cost ${amount}",
        "Spent ${amount} on food at {store}",
        "Ordered {item} online for ${amount}",
        "Dinner at {store} with total bill ${amount}"
    ],
    "Clothing": [
        "Bought new {item} for ${amount}",
        "Shopping at {store}: {item}, ${amount}",
        "Purchased clothes and {item}",
        "Fashion expense on {item} at {store}"
    ],
    "Healthcare": [
        "Bought medicines worth ${amount}",
        "Visited {store} clinic, consultation fee ${amount}",
        "Medical expense: {item} for ${amount}",
        "Spent ${amount} at pharmacy for {item}"
    ],
    "Utilities": [
        "Paid ${amount} for electricity bill",
        "Monthly water bill payment ${amount}",
        "Internet recharge for ${amount}",
        "Utility expense: Gas bill ${amount}"
    ]
}
```

## Example item lists per category

```python
ITEMS = {
    "Stationary": ["notebooks", "pencils", "markers", "files", "folders"],
    "Electronics": ["laptop", "headphones", "mouse", "keyboard", "tablet"],
    "Groceries": ["milk", "bread", "eggs", "cheese", "apples"],
    "Travel": ["plane ticket", "train ticket", "bus ticket", "taxi ride"],
    "Food": ["burger", "pizza", "pasta", "salad", "sushi"],
    "Clothing": ["shirt", "jeans", "jacket", "dress", "shoes"],
    "Healthcare": ["painkillers", "bandages", "vitamins", "antibiotics"],
    "Utilities": ["electricity", "internet", "water", "gas"]
}
```

## Generate a single synthetic transaction description for a given category.

```python
def generate_description(category: str):
    template = random.choice(CATEGORY_TEMPLATES[category])
    item = random.choice(ITEMS[category])
    amount = round(random.uniform(5, 500), 2)
    city = fake.city()
    store = fake.company()
    return template.format(item=item, amount=amount, city=city, store=store)
```

## Generate synthetic transaction data with text description and category.

```python
def generate_synthetic_data(n_samples=5000, categories=None):
    if categories is None:
        categories = list(CATEGORY_TEMPLATES.keys())
    records = []
    for _ in range(n_samples):
        cat = random.choice(categories)
        text = generate_description(cat)
        records.append({"text_description": text, "category": cat})
    return pd.DataFrame(records)
```

## Generate synthetic dataset

```python
df = generate_synthetic_data(n_samples=5000)
```

## Save to CSV

```python
df.to_csv("Synthetic_Data.csv", index=False)
```
